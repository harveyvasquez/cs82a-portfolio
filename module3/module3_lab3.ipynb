```python
import pandas as pd
df = pd.read_csv("bridges.csv")
print(df.shape)
df.head()
```

```python
df["length"].describe().round(2)
```

Assumption: a bridge length is plausible if it is positive and no more than 3500. I chose 3500 because it is the usual outlier limit, Q3 + 1.5 x IQR = 2000 + 1.5 x 1000.

```python
df[(df["length"] <= 0) | (df["length"] > 3500)]
```

Three bridges are flagged: E34 (4558), E46 (4000), and E91 (3756). They are extreme but not impossible, because a long bridge can be thousands of feet. I kept them and marked them in a new column so they can be filtered out later.

```python
df["length_outlier"] = df["length"] > 3500
df[df["length_outlier"]][["Id", "purpose", "length", "length_outlier"]]
```
