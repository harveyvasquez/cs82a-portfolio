```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
df = pd.read_csv("truck.csv")
print(df.shape)
df.head()
```

Correlation

```python
r = df["X"].corr(df["Y"])
print("r =", round(r, 4))
```

The correlation is r = 0.9985, which is almost perfect. Trucks with a higher list price almost always have a higher best price.

Scatter plot

```python
df.plot.scatter(x="X", y="Y")
plt.title("Truck best price rises almost in line with list price (r = 0.9985)")
plt.xlabel("List price (thousands of dollars)")
plt.ylabel("Best price (thousands of dollars)")
plt.show()
```

The points fall almost on a straight line going up, which agrees with r = 0.9985.

Prediction

```python
slope, intercept = np.polyfit(df["X"], df["Y"], 1)
list_price = 20
predicted = slope * list_price + intercept
print("List price:", list_price, "thousand dollars")
print("Predicted best price:", round(predicted, 2), "thousand dollars")
```

I predict that a truck with a list price of 20 thousand dollars has a best price of about 17.46 thousand dollars ($17,457). I am fairly confident because r is 0.9985 and 20 is inside the list prices in the data (12.4 to 22.4). I would trust it much less for a list price far outside that range, and with only 23 trucks the prediction is an estimate, not a guarantee.
