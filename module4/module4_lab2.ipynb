```python
import pandas as pd
df = pd.read_csv("allbacks.csv")
print(df.shape)
df.head()
```

```python
sorted_cover = df["cover"].sort_values().reset_index(drop=True)
print("Median cover:", sorted_cover[len(sorted_cover) // 2])
```

The median cover type is pb. When the 15 covers are sorted, the middle one (the 8th) is a paperback, which means paperbacks are the majority of the books (8 out of 15).

```python
print(df["weight"].mean())
```

The mean weight of all 15 books is about 706.67 grams, so a typical book in this sample weighs a little over 700 grams.

```python
df.groupby("cover")["weight"].mean()
```

Hardback books average 796.43 grams and paperbacks average 628.13 grams. Hardbacks are about 168 grams heavier, so switching books to softback would make them lighter to store and ship. With only 15 books this is a rough guide.

By hand check: 800 + 950 + 1050 + 350 + 750 + 600 + 1075 + 250 + 700 + 650 + 975 + 350 + 950 + 425 + 725 = 10600, and 10600 / 15 = 706.67. Python gives the same sum and mean below.
