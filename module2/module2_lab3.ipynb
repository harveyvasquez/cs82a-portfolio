Temperature cutoffs (degrees Celsius)

Cold: below 15. 
Warm: 15 to 25, including both 15 and 25. 
Hot: above 25.

```python
temperatures = [10, 16, 23, 28, 14, 22, 30]

for temp in temperatures:
    if temp < 15:
        label = "Cold"
    elif temp <= 25:
        label = "Warm"
    else:
        label = "Hot"
    print(temp, "is", label)
```

```python
boundary_tests = [15, 25, 14, 26]

for temp in boundary_tests:
    if temp < 15:
        label = "Cold"
    elif temp <= 25:
        label = "Warm"
    else:
        label = "Hot"
    print(temp, "is", label)
```

The boundary tests show that exactly 15 and 25 are both warm, as my cutoffs say.
