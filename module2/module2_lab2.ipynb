```python
def multiply_for(a, b):
    result = 0
    for i in range(b):
        result = result + a
    return result

print(multiply_for(7, 6))
```

```python
def multiply_while(a, b):
    result = 0
    count = 0
    while count < b:
        result = result + a
        count = count + 1
    return result

print(multiply_while(7, 6))
```

each line below shows three numbers: the result from multiply_for, the result from multiply_while, and the real product using the * operator. The three numbers match on every line, which shows both functions give the correct product.

```python
print(multiply_for(7, 6), multiply_while(7, 6), 7 * 6)
print(multiply_for(12, 12), multiply_while(12, 12), 12 * 12)
print(multiply_for(0, 5), multiply_while(0, 5), 0 * 5)
print(multiply_for(25, 4), multiply_while(25, 4), 25 * 4)
```

Understanding Shown:
The while loop stops because count goes up by 1 each time the loop runs. Once count reaches b, the test "count < b" is false, so the loop ends. Without the line "count = count + 1", count would never change and the loop would never stop.
